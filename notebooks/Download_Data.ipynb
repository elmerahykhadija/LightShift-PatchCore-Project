## Étape 1 — Installation des dépendances

```python
# Installation si pas déjà dans le container
import subprocess, sys

def pip_install(pkg):
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])

try:
    import kagglehub
    print("✅ kagglehub déjà installé")
except ImportError:
    pip_install("kagglehub")
    import kagglehub
    print("✅ kagglehub installé")

try:
    import anomalib
    print("✅ anomalib déjà installé")
except ImportError:
    pip_install("git+https://github.com/open-edge-platform/anomalib.git")
    print("✅ anomalib installé")
```

## Étape 2 — Configuration des chemins

```python
from pathlib import Path

# Chemin de destination dans le volume Docker
DATA_DIR = Path("/workspace/data/mvtec")
DATA_DIR.mkdir(parents=True, exist_ok=True)

print(f"📁 Dossier de destination : {DATA_DIR}")
print(f"📂 Contenu actuel : {list(DATA_DIR.iterdir()) if DATA_DIR.exists() else 'vide'}")
```

## Étape 3 — Téléchargement du dataset MVTec AD depuis Kaggle

> ⚠️ Assurez-vous que vos credentials Kaggle sont configurés :
> - Placez `kaggle.json` dans `~/.kaggle/kaggle.json` dans le container
> - Ou définissez les variables d'env `KAGGLE_USERNAME` et `KAGGLE_KEY`

```python
import kagglehub

print("⬇️ Téléchargement du dataset MVTec AD...")
path = kagglehub.dataset_download("ipythonx/mvtec-ad")
path = Path(path)

print(f"✅ Dataset téléchargé dans : {path}")
print(f"📂 Contenu : {[p.name for p in path.iterdir()]}")
```

## Étape 4 — Copie vers `/workspace/data/mvtec`

```python
import shutil
from pathlib import Path

source = Path(path)  # chemin du cache kagglehub
destination = Path("/workspace/data/mvtec")
destination.mkdir(parents=True, exist_ok=True)

categories = [
    "bottle", "cable", "capsule", "carpet", "grid",
    "hazelnut", "leather", "metal_nut", "pill",
    "screw", "tile", "toothbrush", "transistor",
    "wood", "zipper"
]

for category in categories:
    src = source / category
    dst = destination / category
    if src.exists():
        print(f"📋 Copie de {category}...")
        shutil.copytree(src, dst, dirs_exist_ok=True)
        print(f"  ✅ {category} : OK")
    else:
        print(f"  ⚠️ {category} : non trouvé dans {src}")

print("\n🎉 Toutes les catégories ont été sauvegardées !")
print(f"📁 Disponible dans : {destination}")
```

## Étape 5 — Vérification

```python
from pathlib import Path

data_dir = Path("/workspace/data/mvtec")

print("📊 Vérification du dataset MVTec AD :\n")
for cat_dir in sorted(data_dir.iterdir()):
    if cat_dir.is_dir():
        n_train = len(list((cat_dir / "train" / "good").glob("*"))) if (cat_dir / "train" / "good").exists() else 0
        n_test  = len(list((cat_dir / "test").rglob("*.png"))) + len(list((cat_dir / "test").rglob("*.jpg"))) if (cat_dir / "test").exists() else 0
        print(f"  🗂️  {cat_dir.name:<15} | train/good: {n_train:>4}  | test images: {n_test:>4}")
```
