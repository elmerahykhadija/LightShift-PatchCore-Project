```python
import shutil
from pathlib import Path

source = Path(
    "/root/.cache/kagglehub/datasets/ipythonx/mvtec-ad/versions/2"
)

destination = Path("/content/drive/MyDrive/data/mvtec")
destination.mkdir(parents=True, exist_ok=True)

categories = [
    "bottle", "cable", "capsule", "carpet", "grid",
    "hazelnut", "leather", "metal_nut", "pill",
    "screw", "tile", "toothbrush", "transistor",
    "wood", "zipper"
]

for category in categories:
    print(f"Copie de {category}...")

    shutil.copytree(
        source / category,
        destination / category,
        dirs_exist_ok=True
    )

    print(f"{category} : OK")

print("Toutes les catégories ont été sauvegardées !")
```
